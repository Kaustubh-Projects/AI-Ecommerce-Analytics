```python

# Load the geolocation dataset
geolocation_file = (
    project_folder
    / "data"
    / "raw"
    / "olist_geolocation_dataset.csv"
)

geolocation = pd.read_csv(geolocation_file)

# Check the dataset size and ZIP-prefix uniqueness
print("Geolocation rows:", len(geolocation))
print(
    "Unique geolocation ZIP prefixes:",
    geolocation["geolocation_zip_code_prefix"].nunique()
)
print(
    "Duplicate geolocation ZIP-prefix rows:",
    geolocation["geolocation_zip_code_prefix"].duplicated().sum()
)

# Compare customer and seller ZIP prefixes with geolocation
customer_zips = set(
    customers["customer_zip_code_prefix"].dropna().unique()
)
seller_zips = set(
    sellers["seller_zip_code_prefix"].dropna().unique()
)
geolocation_zips = set(
    geolocation["geolocation_zip_code_prefix"].dropna().unique()
)

print(
    "\nCustomer ZIP prefixes without geolocation:",
    len(customer_zips - geolocation_zips)
)
print(
    "Seller ZIP prefixes without geolocation:",
    len(seller_zips - geolocation_zips)
)
```

```python

# Count distinct coordinate pairs for each ZIP prefix
locations_per_zip = geolocation.groupby(
    "geolocation_zip_code_prefix"
).apply(
    lambda group: group[
        ["geolocation_lat", "geolocation_lng"]
    ].drop_duplicates().shape[0]
)

# Count ZIP prefixes associated with multiple coordinate pairs
multiple_coordinate_zips = locations_per_zip[locations_per_zip > 1]

print(
    "ZIP prefixes with multiple coordinate pairs:",
    len(multiple_coordinate_zips)
)

print(
    "Maximum coordinate pairs for one ZIP prefix:",
    locations_per_zip.max()
)

# Display five examples for inspection
print("\nExample ZIP prefixes with multiple coordinate pairs:")
display(multiple_coordinate_zips.head())
```
