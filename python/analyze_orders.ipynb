```python
from pathlib import Path
import pandas as pd

# Find the project folder
project_folder = Path.cwd().parent

# Find the orders CSV file
file_path = project_folder / "data" / "raw" / "olist_orders_dataset.csv"

print("Dataset path:", file_path)
print("Pandas version:", pd.__version__)
```

```python
orders = pd.read_csv(file_path)

print("Dataset loaded successfully!")
```

```python
missing_delivery = orders[
    (orders["order_status"] == "delivered")
    & (orders["order_delivered_customer_date"].isnull())
]

print("Delivered orders without delivery dates:", len(missing_delivery))
```

```python
# Count missing customer delivery dates by order status
orders.groupby("order_status")[
    "order_delivered_customer_date"
].apply(lambda column: column.isnull().sum())
```
